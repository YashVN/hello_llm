```python
from langchain.messages import HumanMessage, AIMessage, SystemMessage
```

```python
def print_response(response):
    for block in response.content_blocks:
        if block['type'] == 'text':
            print(block['text'])
```

```python
messages = [
    HumanMessage("what is 2 * 2 + 3 ?"),
    AIMessage("""
    step 1: 2 * 2 = 4
    step 2: 4 + 3 = 7

    Answer: 7
    """),
    HumanMessage("What is 3 - 3 + 6 + 4 ?"),
    AIMessage("""
    step 1: 3 - 3 = 0
    step 2: 0 + 6 = 6
    step 3: 6 + 4 = 10

    Answer: 10
    """),
    HumanMessage("what is 5 * 6 - 2 ?")
    
]

response = llm.invoke(messages)
print(response)
```

```python
response = llm.invoke(messages)
print_response(response)
```

Creating Agent

```python
#creating an Agent 
from langchain.agents import create_agent
```

```python
#Approach 1
model = init_chat_model(model="google_genai:gemini-3.1-flash-lite")
agent = create_agent(
    model=model,
    tools=[]
)
response = agent.invoke({
    "messages": ["What is the capital of France?"]
})
for message in response['messages']:
    print(message)
```

```python
#Approach 2
agent = create_agent(
    model = "google_genai:gemini-3.1-flash-lite",
    tools = []
)
reponse = agent.invoke({
    "messages": ["What is the capital of France?"]
})
for message in response['messages']:
    print(message)
```
